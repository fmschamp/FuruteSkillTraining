# Exercise 1 — Exploratory Data Analysis & Automated Reporting with Pandas

## Objective
Generate a reproducible numerical dataset, visualize the relationship between `f2` and `f3`, calculate the row-wise mean in `f5`, and export the complete table to Excel.

## Logic
- A fixed random seed makes the sample data reproducible.
- `DataFrame.mean(axis=1)` calculates the mean across columns for each row.
- The Excel output is saved in the same folder as this notebook.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_DIR = Path(r"C:\Users\fmsch\assignment")
OUTPUT_DIR = PROJECT_DIR / "Exercise1_DhruvRawat"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Generate 100 rows of random numerical values in four columns.
np.random.seed(42)
df = pd.DataFrame({
    "f1": np.random.uniform(10, 100, 100),
    "f2": np.random.uniform(10, 100, 100),
    "f3": np.random.uniform(10, 100, 100),
    "f4": np.random.uniform(10, 100, 100)
})

# Add the arithmetic mean of f1, f2, f3 and f4 for every row.
df["f5"] = df[["f1", "f2", "f3", "f4"]].mean(axis=1)

display(df.head())
print("Dataset shape:", df.shape)
```

```python
# Scatter plot: f2 on the X-axis and f3 on the Y-axis.
plt.figure(figsize=(8, 5))
plt.scatter(df["f2"], df["f3"], alpha=0.7)
plt.xlabel("f2")
plt.ylabel("f3")
plt.title("Relationship between f2 and f3")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "f2_vs_f3_scatter.png", dpi=150)
plt.show()
```

```python
# Export all five columns to an Excel workbook.
excel_path = OUTPUT_DIR / "Exercise1_DhruvRawat_Output.xlsx"
df.to_excel(excel_path, index=False, sheet_name="Generated_Data")
print(f"Excel file saved to: {excel_path.resolve()}")
```
