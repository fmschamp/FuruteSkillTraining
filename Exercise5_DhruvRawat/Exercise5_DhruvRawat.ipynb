# Exercise 5 — Interactive AI Prototyping with Gradio
## Objective
Create a browser-based conversational interface using Gradio's `ChatInterface`.

This implementation is a lightweight demonstration chatbot. It uses a rule-based response function so it can run without an API key or paid language-model service. It demonstrates the interactive UI; it is not a full generative AI model.

```python
from pathlib import Path
PROJECT_DIR = Path(r"C:\Users\fmsch\assignment")
OUTPUT_DIR = PROJECT_DIR / "Exercise5_DhruvRawat"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print("Exercise 5 output folder:", OUTPUT_DIR)
```

```python
import gradio as gr
from datetime import datetime

def respond(message, history):
    """Return a simple demonstration response based on the user's message."""
    text = message.strip()
    lower = text.lower()

    if not text:
        return "Please enter a message so I can respond."

    if any(word in lower for word in ["hello", "hi", "namaste"]):
        return "Hello! Welcome to the AI Training Demo Chatbot. How can I help you today?"
    if "pandas" in lower or "dataframe" in lower:
        return "Pandas is a Python library for data analysis. A DataFrame stores tabular data in rows and columns."
    if "artificial intelligence" in lower or lower == "what is ai?":
        return "Artificial Intelligence (AI) is the field of building systems that perform tasks commonly associated with human intelligence, such as understanding language, recognizing patterns and making predictions."
    if "opencv" in lower or "image" in lower:
        return "OpenCV is a computer-vision library. In Python, cv2.imread() loads an image and cv2.addWeighted() can blend two same-sized images."
    return (
        "Thank you for your question. This is a demonstration chatbot interface. "
        "To enable advanced AI-generated answers, connect a permitted LLM API to this response function."
    )
```

```python
demo = gr.ChatInterface(
    fn=respond,
    title="Government Officer Training — AI Chatbot",
    description=(
        "Welcome! This is a simple conversational AI prototype for training. "
        "Enter a question below or try one of the example prompts."
    ),
    examples=[
        "What is Artificial Intelligence?",
        "Explain Pandas DataFrame with an example.",
        "How can OpenCV blend two images?"
    ],
    chatbot=gr.Chatbot(height=420),
    textbox=gr.Textbox(
        placeholder="Type your question here and press Enter...",
        label="Your message"
    )
)

# In Jupyter, Gradio displays an embedded interface and may also print a local URL.
# For a shareable public URL, use share=True only if permitted by your organization.
demo.launch(share=False, inline=True)
```
